```python
import seaborn as sns
import pandas as pd

#Note: Both seaborn and pandas as Python libraries. Libraries are collection of code that contain functions and modules that can be used. Libraries make running Python more efficient, usually in a simple line of code
```

```python
import pandas as pd
url = "https://raw.githubusercontent.com/chiharuishida/Tutorialdata/main/iris.csv"
Iris = pd.read_csv(url)
#Note that Iris is the dataframe I just named
```

```python
Iris.columns=['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm', 'Species']
```

```python
Iris.head(10)
#provides the first 10 raw of data. this provides a quick view of the dataset titled "tips"
```

```python
Iris.describe()
#quickly provides a descriptive statistics for all the continous variables in the dataset
```

```python
Iris.info()
#"info" provides information about each variable in the dataset. Dtype of float is a number with a decimal place, while int is for integers
```

```python
Iris['Species'].value_counts()
#value_counts counts the frequency in the categorical variable. "Iris-Setosa" is a categorical variable in our data
```

```python
Iris.corr(numeric_only = True)
#quickly provides a correlation matrix
```

```python
sns.heatmap(Iris.corr(numeric_only=True), annot=True)
#annot=True adds an annotation to the right
```

```python
sns.heatmap(Iris.corr(numeric_only=True), cmap="YlGnBu")
#just changing colors here for better looks (darker=greater value)
#other colormap options are 'Greys', 'Purples', 'Blues', 'Greens', 'Oranges', 'Reds','YlOrBr', 'YlOrRd', 'OrRd', 'PuRd', 'RdPu', 'BuPu','GnBu', 'PuBu', 'YlGnBu', 'PuBuGn', 'BuGn', 'YlGn'
```

```python
sns.pairplot(Iris)
#quickly provides pairplot between the continous varialbles.
```

```python
sns.scatterplot(x="SepalLengthCm", y="PetalLengthCm", data=Iris)
#let's just look at the scatterplot between "top" and "total_bill"
#A scatterplot is for any two continuous variables
```

```python
sns.lmplot(x="SepalLengthCm", y="PetalLengthCm", data=Iris)
#fitting a trendline (Tableau term for fitting a line in the scatterplot)
```

```python
sns.countplot(x='Species', data=Iris)
#countplot creates a bar chart. the variable could be continuous or categorical
```

```python
sns.countplot(x='SepalWidthCm', data=Iris)
```
